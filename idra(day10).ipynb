```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)
pd.set_option("display.max_rows", 100)
```

```python
# Update this path only if the CSV is moved to another location.
file_path = "Day10_Flight_Operations_Dataset.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully.")
print("Shape:", df.shape)
df.head()
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types and non-null counts:")
df.info()
```

```python
print("First 5 records:")
display(df.head())

print("Last 5 records:")
display(df.tail())
```

```python
# Numerical summary
display(df.describe(include="all").T)
```

```python
missing = df.isnull().sum().sort_values(ascending=False)
missing_pct = (df.isnull().mean() * 100).sort_values(ascending=False)

missing_summary = pd.DataFrame({
    "Missing_Count": missing,
    "Missing_Percentage": missing_pct.round(2)
})

display(missing_summary)

print("Duplicate rows:", df.duplicated().sum())
```

```python
categorical_cols = df.select_dtypes(include=["object", "category"]).columns

for col in categorical_cols:
    print(f"\n--- {col} ---")
    print("Unique values:", df[col].nunique(dropna=True))
    display(df[col].value_counts(dropna=False).head(15))
```

```python
# Select all numeric columns
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
print("Numeric columns:", numeric_cols)

if numeric_cols:
    print("\nSelected numeric columns:")
    display(df[numeric_cols].head())
```

```python
# First 5 rows using iloc
display(df.iloc[:5])

# First 5 rows and first 3 columns
display(df.iloc[:5, :3])

# First 5 rows using loc
display(df.loc[df.index[:5]])
```

```python
# Example 1: filter rows where at least one numeric value is above its median.
# This demonstrates boolean filtering in a dataset-independent way.
if numeric_cols:
    filter_col = numeric_cols[0]
    threshold = df[filter_col].median()
    filtered = df[df[filter_col] > threshold]
    print(f"Filtering {filter_col} > median ({threshold}):")
    display(filtered.head(10))
else:
    print("No numeric columns are available for numeric filtering.")
```

```python
# Example 2: filter a categorical column by its most common category.
if len(categorical_cols) > 0:
    filter_col = categorical_cols[0]
    common_value = df[filter_col].mode(dropna=True).iloc[0]
    filtered_cat = df[df[filter_col] == common_value]
    print(f"Filtering {filter_col} == {common_value!r}:")
    display(filtered_cat.head(10))
else:
    print("No categorical columns are available for categorical filtering.")
```

```python
if numeric_cols:
    sort_col = numeric_cols[0]
    print(f"Top records by {sort_col}:")
    display(df.sort_values(by=sort_col, ascending=False).head(10))

    print(f"Bottom records by {sort_col}:")
    display(df.sort_values(by=sort_col, ascending=True).head(10))
```

```python
group_candidates = [
    c for c in categorical_cols
    if 1 < df[c].nunique(dropna=True) <= 30
]

if group_candidates and numeric_cols:
    group_col = group_candidates[0]
    agg_dict = {c: ["count", "mean", "min", "max"] for c in numeric_cols[:4]}

    grouped = df.groupby(group_col, dropna=False).agg(agg_dict)
    display(grouped)
else:
    print("Not enough suitable categorical/numeric columns for automatic aggregation.")
```

```python
if group_candidates:
    count_col = group_candidates[0]
    count_summary = (
        df.groupby(count_col, dropna=False)
          .size()
          .reset_index(name="Flight_Count")
          .sort_values("Flight_Count", ascending=False)
    )
    display(count_summary)
```

```python
# Create a copy before transformations
analysis_df = df.copy()

# Convert likely date/time columns to datetime when possible.
date_like_cols = [
    c for c in analysis_df.columns
    if any(key in c.lower() for key in ["date", "time", "day"])
]

converted_date_cols = []
for col in date_like_cols:
    converted = pd.to_datetime(analysis_df[col], errors="coerce")
    # Convert only if a reasonable number of values can be parsed.
    if converted.notna().sum() > 0:
        analysis_df[col] = converted
        converted_date_cols.append(col)

print("Converted date/time-like columns:", converted_date_cols)
```

```python
# Add simple derived features from the first usable datetime column.
if converted_date_cols:
    date_col = converted_date_cols[0]
    analysis_df["Analysis_Month"] = analysis_df[date_col].dt.month
    analysis_df["Analysis_Month_Name"] = analysis_df[date_col].dt.month_name()
    analysis_df["Analysis_Day_Name"] = analysis_df[date_col].dt.day_name()

    display(analysis_df[[date_col, "Analysis_Month_Name", "Analysis_Day_Name"]].head())
else:
    print("No usable date column was detected.")
```

```python
if numeric_cols:
    transform_col = numeric_cols[0]
    values = analysis_df[transform_col].dropna()

    if values.nunique() >= 3:
        analysis_df[f"{transform_col}_Band"] = pd.qcut(
            analysis_df[transform_col],
            q=3,
            labels=["Low", "Medium", "High"],
            duplicates="drop"
        )
        display(analysis_df[[transform_col, f"{transform_col}_Band"]].head(10))
    else:
        print(f"{transform_col} does not have enough unique values for banding.")
```

<h4>Question 1 — Which category has the most flight records?</h4>

```python
if group_candidates:
    category_col = group_candidates[0]
    category_counts = (
        analysis_df[category_col]
        .value_counts(dropna=False)
        .rename_axis(category_col)
        .reset_index(name="Flight_Count")
    )
    display(category_counts.head(10))
    print("Most frequent category:", category_counts.iloc[0][category_col])
```

<h4>Question 2 — Which category has the highest average numeric measure?</h4>

```python
if group_candidates and numeric_cols:
    category_col = group_candidates[0]
    measure_col = numeric_cols[0]

    avg_by_category = (
        analysis_df.groupby(category_col, dropna=False)[measure_col]
        .mean()
        .sort_values(ascending=False)
        .reset_index(name=f"Average_{measure_col}")
    )
    display(avg_by_category.head(10))
```
