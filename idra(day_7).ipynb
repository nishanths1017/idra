```python
missing_count = df.isnull().sum()
missing_percentage = (df.isnull().mean() * 100).round(2)

missing_summary = pd.DataFrame({
    'Missing Values': missing_count,
    'Missing Percentage': missing_percentage
})

missing_summary = missing_summary[missing_summary['Missing Values'] > 0]

if missing_summary.empty:
    print("No missing values were detected.")
else:
    display(missing_summary.sort_values('Missing Values', ascending=False))
```

```python
duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns.tolist()
categorical_columns = df.select_dtypes(exclude=np.number).columns.tolist()
missing_total = int(df.isnull().sum().sum())

print("DATASET SUMMARY")
print("=" * 50)
print(f"Rows                 : {df.shape[0]}")
print(f"Columns              : {df.shape[1]}")
print(f"Numeric features     : {len(numeric_columns)}")
print(f"Categorical features : {len(categorical_columns)}")
print(f"Missing values       : {missing_total}")
print(f"Duplicate rows       : {duplicate_count}")
print("\nAvailable features:")
for column in df.columns:
    print(f"- {column} ({df[column].dtype})")
```

```python
print("BASIC OBSERVATIONS")
print("=" * 50)
print(f"1. The dataset contains {df.shape[0]:,} rows and {df.shape[1]:,} columns.")
print(f"2. There are {len(numeric_columns)} numerical feature(s) and {len(categorical_columns)} non-numerical feature(s).")
print(f"3. The dataset contains {missing_total:,} missing value(s) in total.")
print(f"4. The dataset contains {duplicate_count:,} duplicate row(s).")

if missing_total > 0:
    affected = missing_summary.index.tolist()
    print("5. Columns with missing values:", ", ".join(affected))
else:
    print("5. No columns contain missing values.")

print("\nFeature names:")
print(", ".join(df.columns.astype(str)))
```
