```python
import pandas as pd
import numpy as np
from IPython.display import display, Markdown

file_path = 'Day16_Student_Wellbeing_Survey.csv'  # Change path if using Google Drive/Colab
df = pd.read_csv(file_path)

print("Dataset Shape:", df.shape)
display(df.head())
print("\nColumn Names:")
print(df.columns.tolist())
print("\nData Types:")
display(df.dtypes.to_frame("Data Type"))
print("\nMissing Values:")
display(df.isnull().sum().to_frame("Missing Values"))
```

```python
variables = [
    'Weekly_Study_Hours',
    'Average_Sleep_Hours',
    'Daily_Screen_Time_Hours',
    'Stress_Score',
    'Academic_Readiness_Score'
]

descriptive = pd.DataFrame(index=variables)
descriptive['Mean'] = df[variables].mean()
descriptive['Median'] = df[variables].median()
descriptive['Mode'] = [df[v].mode().iloc[0] for v in variables]
descriptive['Range'] = df[variables].max() - df[variables].min()
descriptive['Variance'] = df[variables].var()
descriptive['Standard Deviation'] = df[variables].std()
descriptive['Q1'] = df[variables].quantile(0.25)
descriptive['Q3'] = df[variables].quantile(0.75)
descriptive['IQR'] = descriptive['Q3'] - descriptive['Q1']

display(descriptive.round(4))
```

```python
greatest_variability = descriptive['Variance'].idxmax()

print(f"Variable with greatest variability (based on variance): {greatest_variability}")
print(f"Variance = {descriptive.loc[greatest_variability, 'Variance']:.4f}")
print(f"Standard Deviation = {descriptive.loc[greatest_variability, 'Standard Deviation']:.4f}")

print("\nInterpretation:")
print(f"{greatest_variability} has the greatest variability because it has the largest variance and standard deviation among the selected variables.")
```

```python
outlier_variables = [
    'Weekly_Study_Hours',
    'Daily_Screen_Time_Hours',
    'Commute_Time_Minutes',
    'Monthly_Discretionary_Spending'
]

outlier_summary = []

for variable in outlier_variables:
    q1 = df[variable].quantile(0.25)
    q3 = df[variable].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = df[(df[variable] < lower_bound) | (df[variable] > upper_bound)][variable]

    outlier_summary.append({
        'Variable': variable,
        'Q1': q1,
        'Q3': q3,
        'IQR': iqr,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Number of Outliers': len(outliers)
    })

outlier_df = pd.DataFrame(outlier_summary)
display(outlier_df.round(4))
```

```python
print("Outlier Interpretation:")
for variable in outlier_variables:
    q1 = df[variable].quantile(0.25)
    q3 = df[variable].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outliers = df[(df[variable] < lower_bound) | (df[variable] > upper_bound)][variable]

    print(f"\n{variable}: {len(outliers)} outlier(s)")
    if len(outliers) > 0:
        print("Outlier values:", outliers.round(2).tolist())
```

```python
variable = 'Daily_Screen_Time_Hours'

q1 = df[variable].quantile(0.25)
q3 = df[variable].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

clean_data = df[(df[variable] >= lower_bound) & (df[variable] <= upper_bound)][variable]

comparison = pd.DataFrame({
    'Dataset': ['Before Removing Outliers', 'After Removing Outliers'],
    'Mean': [df[variable].mean(), clean_data.mean()],
    'Median': [df[variable].median(), clean_data.median()],
    'Number of Records': [len(df), len(clean_data)]
})

display(comparison.round(4))

print("Interpretation:")
print("If the mean changes more than the median after removing outliers, it indicates that the mean is more sensitive to extreme values.")
```

```python
A = df['Part_Time_Job'].astype(str).str.strip().eq('Yes')
B = df['Stress_Score'].ge(7)
C = df['Scholarship'].astype(str).str.strip().eq('Yes')
D = df['Exercise_Days_Per_Week'].ge(3)

P_A = A.mean()
P_B = B.mean()
P_C = C.mean()
P_D = D.mean()
P_A_or_B = (A | B).mean()
P_A_and_B = (A & B).mean()
P_A_given_B = (A & B).sum() / B.sum()
P_B_given_A = (A & B).sum() / A.sum()

probabilities = pd.DataFrame({
    'Probability': ['P(A)', 'P(B)', 'P(C)', 'P(D)', 'P(A or B)', 'P(A and B)', 'P(A|B)', 'P(B|A)'],
    'Value': [P_A, P_B, P_C, P_D, P_A_or_B, P_A_and_B, P_A_given_B, P_B_given_A]
})

probabilities['Percentage'] = probabilities['Value'] * 100
display(probabilities.round(4))

print("\nEvent Counts:")
print("A:", A.sum(), "| B:", B.sum(), "| C:", C.sum(), "| D:", D.sum(), "| A and B:", (A & B).sum())
```

```python
year1 = df['Year_of_Study'].eq(1)
year4 = df['Year_of_Study'].eq(4)
intersection = (year1 & year4).sum()

print("Students in Year 1:", year1.sum())
print("Students in Year 4:", year4.sum())
print("Students simultaneously in Year 1 and Year 4:", intersection)

if intersection == 0:
    print("\nConclusion: Yes, Year_of_Study = 1 and Year_of_Study = 4 are mutually exclusive events.")
    print("A single student cannot belong to both Year 1 and Year 4 at the same time.")
else:
    print("\nConclusion: No, the events are not mutually exclusive.")
```

```python
product_P_A_P_B = P_A * P_B

print(f"P(A and B) = {P_A_and_B:.6f}")
print(f"P(A) × P(B) = {product_P_A_P_B:.6f}")

if np.isclose(P_A_and_B, product_P_A_P_B):
    print("\nConclusion: A and B appear independent because the probabilities are approximately equal.")
else:
    print("\nConclusion: A and B do NOT appear independent because P(A and B) differs from P(A) × P(B).")
    print("In this dataset, having a part-time job and having a high stress score show statistical dependence.")
```

```python
not_A = ~A

P_not_A = not_A.mean()
P_B_given_not_A = (B & not_A).sum() / not_A.sum()

P_B_from_total_probability = (
    P_B_given_A * P_A +
    P_B_given_not_A * P_not_A
)

P_A_given_B_bayes = (
    P_B_given_A * P_A
) / P_B_from_total_probability

print(f"P(B|A) = {P_B_given_A:.6f}")
print(f"P(B|not A) = {P_B_given_not_A:.6f}")
print(f"P(A) = {P_A:.6f}")
print(f"P(not A) = {P_not_A:.6f}")
print(f"P(B) from total probability = {P_B_from_total_probability:.6f}")
print(f"\nP(A|B) using Bayes' theorem = {P_A_given_B_bayes:.6f}")
print(f"P(A|B) using direct conditional probability = {P_A_given_B:.6f}")
print(f"Difference = {abs(P_A_given_B_bayes - P_A_given_B):.12f}")

print("\nVerification:")
if np.isclose(P_A_given_B_bayes, P_A_given_B):
    print("The Bayes' theorem result agrees with the direct conditional-probability calculation.")
else:
    print("The two results differ; check the event definitions or calculations.")
```

```python
readiness = df['Academic_Readiness_Score']

mean_readiness = readiness.mean()
std_readiness = readiness.std()

highest_score = readiness.max()
lowest_score = readiness.min()

z_highest = (highest_score - mean_readiness) / std_readiness
z_lowest = (lowest_score - mean_readiness) / std_readiness

print(f"Mean Academic_Readiness_Score = {mean_readiness:.4f}")
print(f"Standard Deviation = {std_readiness:.4f}")

print(f"\nHighest Score = {highest_score:.2f}")
print(f"Z-score of Highest Score = {z_highest:.4f}")

print(f"\nLowest Score = {lowest_score:.2f}")
print(f"Z-score of Lowest Score = {z_lowest:.4f}")

print("\nInterpretation:")
print(f"The highest score is {z_highest:.2f} standard deviations above the mean.")
print(f"The lowest score is {abs(z_lowest):.2f} standard deviations below the mean.")
```

```python
empirical_rule = pd.DataFrame({
    'Interval': [
        'Mean ± 1 Standard Deviation',
        'Mean ± 2 Standard Deviations',
        'Mean ± 3 Standard Deviations'
    ],
    'Expected Percentage': ['68%', '95%', '99.7%'],
    'Expected Number of Students (n=600)': [
        0.68 * len(df),
        0.95 * len(df),
        0.997 * len(df)
    ]
})

display(empirical_rule.round(2))

for k in [1, 2, 3]:
    lower = mean_readiness - k * std_readiness
    upper = mean_readiness + k * std_readiness
    print(f"{k} SD interval: {lower:.2f} to {upper:.2f}")
```
